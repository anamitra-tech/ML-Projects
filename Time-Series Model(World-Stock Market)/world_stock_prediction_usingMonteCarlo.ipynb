```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------
CSV_PATH = "/content/World-Stock-Prices-Dataset.csv"
FEATURES = ["Open", "High", "Low", "Close", "Volume"]
WINDOW_SIZE = 20
HIDDEN_SIZE = 64
EPOCHS = 20
BUY_THRESHOLD = 0.55
SELL_THRESHOLD = 0.45
MC_SAMPLES = 100   # Monte Carlo runs

# ------------------------------------------------------------
# LSTM MODEL WITH DROPOUT
# ------------------------------------------------------------
class StockLSTM(nn.Module):
    def __init__(self, input_size, hidden_size, dropout=0.3):
        super(StockLSTM, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out, _ = self.lstm(x)
        out = self.dropout(out[:, -1, :])
        return self.sigmoid(self.fc(out))

# ------------------------------------------------------------
# LOAD + PREPROCESS
# ------------------------------------------------------------
df = pd.read_csv(CSV_PATH)
df["Date"] = pd.to_datetime(df["Date"], errors="coerce", utc=True)
df = df.sort_values(["Ticker", "Date"]).reset_index(drop=True)
df[FEATURES] = df[FEATURES].apply(pd.to_numeric, errors="coerce")

df["Target"] = (df.groupby("Ticker")["Close"].pct_change().shift(-1) > 0).astype(int)
df = df.dropna(subset=FEATURES + ["Target"])

for ticker, g in df.groupby("Ticker"):

    if len(g) < 1000:
        continue

    print(f"\n===== {ticker} | Monte Carlo LSTM =====")

    scaler = StandardScaler()
    scaled_data = scaler.fit_transform(g[FEATURES].values)
    target_data = g["Target"].values

    X_seq, y_seq = [], []
    for i in range(len(scaled_data) - WINDOW_SIZE):
        X_seq.append(scaled_data[i : i + WINDOW_SIZE])
        y_seq.append(target_data[i + WINDOW_SIZE])

    X_seq = np.array(X_seq)
    y_seq = np.array(y_seq)

    split = int(len(X_seq) * 0.8)
    X_train = torch.tensor(X_seq[:split], dtype=torch.float32)
    X_test = torch.tensor(X_seq[split:], dtype=torch.float32)
    y_train = torch.tensor(y_seq[:split], dtype=torch.float32).view(-1, 1)

    model = StockLSTM(len(FEATURES), HIDDEN_SIZE)
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    criterion = nn.BCELoss()

    # Training
    model.train()
    for epoch in range(EPOCHS):
        optimizer.zero_grad()
        outputs = model(X_train)
        loss = criterion(outputs, y_train)
        loss.backward()
        optimizer.step()

    # ------------------------------------------------------------
    # MONTE CARLO INFERENCE
    # ------------------------------------------------------------
    model.train()  # keep dropout ON

    mc_probs = []

    with torch.no_grad():
        for _ in range(MC_SAMPLES):
            probs = model(X_test).numpy().flatten()
            mc_probs.append(probs)

    mc_probs = np.array(mc_probs)

    mean_probs = mc_probs.mean(axis=0)
    std_probs = mc_probs.std(axis=0)

    signals = np.where(mean_probs > BUY_THRESHOLD, "BUY",
              np.where(mean_probs < SELL_THRESHOLD, "SELL", "HOLD"))

    print(pd.Series(signals).value_counts())

    plt.figure(figsize=(12, 4))
    plt.plot(mean_probs, label="Expected P(BUY)")
    plt.fill_between(
        range(len(mean_probs)),
        mean_probs - std_probs,
        mean_probs + std_probs,
        alpha=0.3,
        label="Uncertainty Band"
    )
    plt.axhline(BUY_THRESHOLD, linestyle="--")
    plt.axhline(SELL_THRESHOLD, linestyle="--")
    plt.title(f"{ticker} | Monte Carlo LSTM Signal Probability")
    plt.legend()
    plt.show()
```
