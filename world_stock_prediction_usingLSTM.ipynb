```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# CONFIG (Kept similar to your MLP config)
# ------------------------------------------------------------
CSV_PATH = "World-Stock-Prices-Dataset.csv"
FEATURES = ["Open", "High", "Low", "Close", "Volume"]
WINDOW_SIZE = 20    # LSTM needs to look back 20 days to see the trend
HIDDEN_SIZE = 64
EPOCHS = 20
BUY_THRESHOLD = 0.55
SELL_THRESHOLD = 0.45

# ------------------------------------------------------------
# LSTM MODEL CLASS
# ------------------------------------------------------------
class StockLSTM(nn.Module):
    def __init__(self, input_size, hidden_size):
        super(StockLSTM, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, batch_first=True)
        self.fc = nn.Linear(hidden_size, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # out: [batch, seq_len, hidden_size]
        out, _ = self.lstm(x)
        # Take the output of the last day in the sequence
        return self.sigmoid(self.fc(out[:, -1, :]))

# ------------------------------------------------------------
# LOAD + PREPROCESS
# ------------------------------------------------------------
df = pd.read_csv(CSV_PATH)
df["Date"] = pd.to_datetime(df["Date"], errors="coerce", utc=True)
df = df.sort_values(["Ticker", "Date"]).reset_index(drop=True)
df[FEATURES] = df[FEATURES].apply(pd.to_numeric, errors="coerce")

# TARGET: NEXT-DAY DIRECTION
df["Target"] = (df.groupby("Ticker")["Close"].pct_change().shift(-1) > 0).astype(int)
df = df.dropna(subset=FEATURES + ["Target"])

results = {}

# ------------------------------------------------------------
# TICKER-WISE TRAINING
# ------------------------------------------------------------
for ticker, g in df.groupby("Ticker"):
    if len(g) < 1000: continue # Need more rows for LSTM windows

    print(f"\n===== {ticker} | LSTM Training =====")

    # 1. Scaling
    scaler = StandardScaler()
    scaled_data = scaler.fit_transform(g[FEATURES].values)
    target_data = g["Target"].values

    # 2. CREATE SEQUENCES (The "Windowing" step)
    X_seq, y_seq = [], []
    for i in range(len(scaled_data) - WINDOW_SIZE):
        X_seq.append(scaled_data[i : i + WINDOW_SIZE])
        y_seq.append(target_data[i + WINDOW_SIZE])

    X_seq, y_seq = np.array(X_seq), np.array(y_seq)

    # Split
    split = int(len(X_seq) * 0.8)
    X_train = torch.tensor(X_seq[:split], dtype=torch.float32)
    X_test = torch.tensor(X_seq[split:], dtype=torch.float32)
    y_train = torch.tensor(y_seq[:split], dtype=torch.float32).view(-1, 1)
    y_test = y_seq[split:]

    # 3. TRAIN
    model = StockLSTM(len(FEATURES), HIDDEN_SIZE)
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    criterion = nn.BCELoss()

    model.train()
    for epoch in range(EPOCHS):
        optimizer.zero_grad()
        outputs = model(X_train)
        loss = criterion(outputs, y_train)
        loss.backward()
        optimizer.step()

    # 4. EVALUATE & PLOT
    model.eval()
    with torch.no_grad():
        test_probs = model(X_test).numpy().flatten()

    signals = np.where(test_probs > BUY_THRESHOLD, "BUY",
              np.where(test_probs < SELL_THRESHOLD, "SELL", "HOLD"))

    print(pd.Series(signals).value_counts())

    plt.figure(figsize=(12, 4))
    plt.plot(test_probs, label="P(BUY)", color="royalblue")
    plt.axhline(BUY_THRESHOLD, color="green", linestyle="--", label="BUY")
    plt.axhline(SELL_THRESHOLD, color="red", linestyle="--", label="SELL")
    plt.title(f"{ticker} | LSTM Automated Signal Probability")
    plt.legend()
    plt.show()
```
