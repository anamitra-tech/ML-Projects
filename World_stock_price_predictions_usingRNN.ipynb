```python
# =============================
# Libraries
# =============================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# =============================
# 1. Load Dataset
# =============================
df = pd.read_csv('/content/World-Stock-Prices-Dataset.csv')
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values(['Brand_Name','Date']).reset_index(drop=True)

# =============================
# 2. Function to Create Sliding Windows
# =============================
def create_sequences(dataset, window):
    X, Y = [], []
    for i in range(len(dataset) - window):
        X.append(dataset[i:(i + window), 0])
        Y.append(dataset[i + window, 0])
    return np.array(X), np.array(Y)

# =============================
# 3. RNN Training Function per Ticker
# =============================
def train_rnn_for_ticker(df_stock, window=10, epochs=30, batch_size=32):
    # Sort by date
    df_stock = df_stock.sort_values('Date')
    raw_prices = df_stock['Close'].values.reshape(-1, 1)

    # Skip if not enough data
    if len(raw_prices) < window + 10:
        return None

    # Scale
    scaler = MinMaxScaler()
    scaled_data = scaler.fit_transform(raw_prices)

    # Train-test split
    train_size = int(len(scaled_data) * 0.8)
    train_data = scaled_data[:train_size]
    test_data = scaled_data[train_size - window:]

    # Create sequences
    X_train, y_train = create_sequences(train_data, window)
    X_test, y_test = create_sequences(test_data, window)

    # Reshape for RNN [samples, timesteps, features]
    X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
    X_test = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

    # Build RNN
    model = Sequential([
        SimpleRNN(50, return_sequences=False, input_shape=(X_train.shape[1], 1)),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mean_squared_error')

    # Train
    model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size, verbose=0)

    # Predict
    test_predict = model.predict(X_test)
    test_predict = scaler.inverse_transform(test_predict)
    y_test_actual = scaler.inverse_transform(y_test.reshape(-1,1))
    test_dates = df_stock['Date'].iloc[-len(y_test_actual):]

    # RMSE
    rmse = np.sqrt(mean_squared_error(y_test_actual, test_predict))

    # Return results
    return {
        'dates': test_dates,
        'actual': y_test_actual.flatten(),
        'predicted': test_predict.flatten(),
        'rmse': rmse
    }

# =============================
# 4. Train and Plot for All Tickers
# =============================
tickers = df['Brand_Name'].unique()
results_all = {}

for ticker in tickers:
    df_stock = df[df['Brand_Name'] == ticker].copy()
    result = train_rnn_for_ticker(df_stock, window=10, epochs=30, batch_size=32)
    if result is None:
        continue
    results_all[ticker] = result

    # Plot
    plt.figure(figsize=(14,5))
    plt.plot(result['dates'], result['actual'], label='Actual Price', color='blue', linewidth=2)
    plt.plot(result['dates'], result['predicted'], label='RNN Prediction', color='red', linestyle='--', linewidth=2)
    plt.title(f'{ticker} Stock Price Prediction using Simple RNN | RMSE: {result["rmse"]:.2f}')
    plt.xlabel('Date')
    plt.ylabel('Closing Price')
    plt.legend()
    plt.grid(True)
    plt.show()
```
