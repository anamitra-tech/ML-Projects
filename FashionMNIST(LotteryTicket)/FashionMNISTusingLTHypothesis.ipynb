```python
# ============================================================
# Fashion MNIST — CNN + Lottery Ticket Hypothesis (LTH)
# Drop this entire file into your Colab after your existing
# notebook cells 1-9 (imports, data loading, CNN definition,
# training, evaluation). Everything below picks up from there.
# ============================================================

import torch
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from tqdm.auto import tqdm
from sklearn.metrics import confusion_matrix, classification_report
import copy

# ── 1. DATA (same as your notebook) ──────────────────────────

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

train_dataset = torchvision.datasets.FashionMNIST(
    root='./data', train=True,  download=True, transform=transform)
test_dataset  = torchvision.datasets.FashionMNIST(
    root='./data', train=False, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader  = DataLoader(test_dataset,  batch_size=64)

CLASSES = ['T-shirt','Trouser','Pullover','Dress','Coat',
           'Sandal',  'Shirt',  'Sneaker', 'Bag',  'Boot']

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"device: {device}")


# ── 2. CNN MODEL (same architecture as your notebook) ─────────
#
# The only addition: a separate `masks` dict that stores a
# binary tensor (0/1) for every weight parameter.
# In forward(), each weight is multiplied by its mask BEFORE
# the computation — pruned weights are permanently zero.

class CNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.conv = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1), nn.ReLU(),
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(64, 128, 3, padding=1), nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.fc = nn.Sequential(
            nn.Linear(128 * 7 * 7, 128), nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(128, 10)
        )

    def forward(self, x):
        x = self.conv(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x


# ── 3. HELPER FUNCTIONS ───────────────────────────────────────

def evaluate(model, loader):
    """Return accuracy on a dataloader."""
    model.eval()
    correct = total = 0
    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            preds = model(images).argmax(dim=1)
            correct += (preds == labels).sum().item()
            total   += labels.size(0)
    return correct / total


def train_one_run(model, epochs=5, lr=1e-3):
    """
    Standard training loop.
    Returns list of per-epoch losses.
    Used for BOTH the dense baseline AND each LTH retrain.
    """
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()
    history   = []

    for epoch in range(epochs):
        model.train()
        epoch_loss = 0
        for images, labels in tqdm(train_loader, desc=f"  epoch {epoch+1}", leave=False):
            images, labels = images.to(device), labels.to(device)
            optimizer.zero_grad()
            loss = criterion(model(images), labels)
            loss.backward()

            # ── KEY LTH DETAIL ──────────────────────────────────
            # After loss.backward() fills .grad, zero out the
            # gradient for every pruned weight BEFORE the optimiser
            # step. This stops the optimiser (Adam) from updating
            # weights that should be permanently frozen at zero.
            if hasattr(model, 'masks') and model.masks:
                for name, param in model.named_parameters():
                    if name in model.masks and param.grad is not None:
                        param.grad.data *= model.masks[name]

            optimizer.step()

            # Re-apply mask to weights directly after update.
            # Adam's momentum can nudge a pruned weight slightly
            # away from zero. This line snaps it back.
            if hasattr(model, 'masks') and model.masks:
                with torch.no_grad():
                    for name, param in model.named_parameters():
                        if name in model.masks:
                            param.data *= model.masks[name]

            epoch_loss += loss.item()

        avg = epoch_loss / len(train_loader)
        history.append(avg)

    return history


# ── 4. LTH CORE FUNCTIONS ────────────────────────────────────

def save_initial_weights(model):
    """
    LTH step 1: snapshot W0 before any training.
    Returns a dict  {param_name: tensor_copy}.
    This dict is NEVER modified after creation.
    """
    return {name: param.data.clone()
            for name, param in model.named_parameters()}


def get_pruning_masks(model):
    """
    Return the current masks dict attached to the model.
    If none exist yet (first call), create all-ones masks.
    """
    if not hasattr(model, 'masks') or not model.masks:
        model.masks = {name: torch.ones_like(param.data)
                       for name, param in model.named_parameters()}
    return model.masks


def prune_by_magnitude(model, prune_fraction):
    """
    LTH step 3: global magnitude pruning.

    - Collect |W| for every weight that is currently ALIVE
      (mask == 1) across ALL layers combined.
    - Find the threshold at the prune_fraction percentile.
    - Mark every weight below that threshold as dead (mask = 0).

    'Global' means all layers compete together — the paper uses
    this rather than per-layer pruning because different layers
    naturally have different weight scales.

    Does NOT zero the weight values here — the mask handles
    that in forward() and the gradient zeroing in backward().
    """
    masks = get_pruning_masks(model)

    # collect absolute values of all currently alive weights
    alive_weights = []
    for name, param in model.named_parameters():
        alive = param.data[masks[name] == 1].abs()
        alive_weights.append(alive)

    all_alive = torch.cat(alive_weights)

    # find the threshold
    threshold = torch.quantile(all_alive, prune_fraction)

    # update masks: kill weights below threshold that are still alive
    for name, param in model.named_parameters():
        newly_dead = (param.data.abs() < threshold) & (masks[name] == 1)
        masks[name][newly_dead] = 0

    # apply masks to weights immediately
    with torch.no_grad():
        for name, param in model.named_parameters():
            param.data *= masks[name]

    # sparsity stats
    total  = sum(m.numel() for m in masks.values())
    pruned = sum((m == 0).sum().item() for m in masks.values())
    return pruned / total   # fraction of weights now dead


def reset_to_init(model, W0):
    """
    LTH step 4: THE lottery ticket step.

    Reset every surviving weight (mask == 1) back to its
    original random value from W0.

    What this is NOT:
      - zeroing surviving weights
      - assigning new random values
      - keeping the trained values Wt

    What this IS:
      - restoring the exact same random number the weight
        started with before any training ever happened
      - leaving masks unchanged

    Why it matters: the initial value W0[i,j] already pointed
    in a useful direction for weights that ended up large.
    Keeping that starting point lets the sparse network retrain
    efficiently from scratch without the baggage from training
    in a 100%-dense world.
    """
    masks = get_pruning_masks(model)
    with torch.no_grad():
        for name, param in model.named_parameters():
            # restore init values only for surviving (mask=1) weights
            # pruned weights (mask=0) stay at zero
            param.data.copy_(W0[name] * masks[name])


def count_sparsity(model):
    masks = get_pruning_masks(model)
    total  = sum(m.numel() for m in masks.values())
    pruned = sum((m == 0).sum().item() for m in masks.values())
    return pruned / total


# ── 5. STEP 0: DENSE BASELINE ────────────────────────────────
#
# Build model, save W0, train fully, measure accuracy.
# This is the target we're trying to match with fewer weights.

print("=" * 55)
print("Step 0 — Dense baseline")
print("=" * 55)

model = CNN().to(device)

# snapshot W0 BEFORE training (crucial — must happen first)
W0 = save_initial_weights(model)

# train dense network
dense_history = train_one_run(model, epochs=5, lr=1e-3)
dense_acc     = evaluate(model, test_loader)

print(f"Dense CNN test accuracy: {dense_acc*100:.2f}%")
print(f"Dense CNN total params:  {sum(p.numel() for p in model.parameters()):,}")


# ── 6. LTH LOOP — ITERATIVE MAGNITUDE PRUNING (IMP) ─────────
#
# Each round:
#   a. prune 20% of remaining alive weights by |W|
#   b. reset survivors to W0
#   c. retrain from scratch with masks fixed
#   d. evaluate

print("\n" + "=" * 55)
print("LTH — Iterative Magnitude Pruning")
print("  prune_rate = 20% per round (compounds)")
print("  rounds     = 5")
print("=" * 55)

PRUNE_PER_ROUND = 0.20
N_ROUNDS        = 5
TRAIN_EPOCHS    = 5

results = [{
    "round":    0,
    "sparsity": 0.0,
    "acc":      dense_acc,
    "params":   sum(p.numel() for p in model.parameters()),
}]

for round_idx in range(1, N_ROUNDS + 1):
    print(f"\nRound {round_idx}")

    # step 3: prune by magnitude
    sparsity = prune_by_magnitude(model, PRUNE_PER_ROUND)
    alive    = sum((m == 1).sum().item() for m in model.masks.values())
    print(f"  after pruning:  sparsity={sparsity*100:.1f}%  alive params={alive:,}")

    # step 4: reset survivors to W0
    reset_to_init(model, W0)
    print(f"  survivors reset to W0")

    # step 5: retrain sparse network from scratch
    print(f"  retraining for {TRAIN_EPOCHS} epochs...")
    train_one_run(model, epochs=TRAIN_EPOCHS, lr=1e-3)

    # evaluate
    acc = evaluate(model, test_loader)
    print(f"  test accuracy:  {acc*100:.2f}%  (dense was {dense_acc*100:.2f}%  Δ={acc-dense_acc:+.4f})")

    results.append({
        "round":    round_idx,
        "sparsity": sparsity,
        "acc":      acc,
        "params":   alive,
    })


# ── 7. RESULTS TABLE ─────────────────────────────────────────

print("\n" + "=" * 55)
print(f"{'Round':>6}  {'Sparsity':>10}  {'Alive params':>14}  {'Test acc':>10}")
print("-" * 48)
for r in results:
    marker = " ← winner" if r["acc"] == max(x["acc"] for x in results) else ""
    print(f"  {r['round']:>4}    {r['sparsity']*100:>8.1f}%    {r['params']:>12,}    {r['acc']*100:>8.2f}%{marker}")


# ── 8. ACCURACY vs SPARSITY PLOT ─────────────────────────────

rounds     = [r["round"]    for r in results]
accuracies = [r["acc"]*100  for r in results]
sparsities = [r["sparsity"]*100 for r in results]

fig, ax1 = plt.subplots(figsize=(8, 4))

ax1.plot(rounds, accuracies, marker='o', color='steelblue', label='Test accuracy (%)')
ax1.axhline(dense_acc * 100, color='steelblue', linestyle='--', alpha=0.4, label='Dense baseline')
ax1.set_xlabel("Pruning round")
ax1.set_ylabel("Test accuracy (%)", color='steelblue')
ax1.tick_params(axis='y', labelcolor='steelblue')

ax2 = ax1.twinx()
ax2.bar(rounds, sparsities, alpha=0.2, color='coral', label='Sparsity (%)')
ax2.set_ylabel("Sparsity (%)", color='coral')
ax2.tick_params(axis='y', labelcolor='coral')

plt.title("LTH on Fashion MNIST — Accuracy vs Sparsity per Round")
fig.tight_layout()
plt.show()


# ── 9. CONFUSION MATRIX OF WINNING TICKET ────────────────────

best_round = max(results[1:], key=lambda r: r["acc"])
print(f"\nWinning ticket: round {best_round['round']}  "
      f"sparsity={best_round['sparsity']*100:.1f}%  "
      f"acc={best_round['acc']*100:.2f}%")

# re-run to that round to get the winning ticket model state
model_ticket = CNN().to(device)
model_ticket.masks = {}
for name, param in model_ticket.named_parameters():
    model_ticket.masks[name] = torch.ones_like(param.data)

for _ in range(best_round["round"]):
    prune_by_magnitude(model_ticket, PRUNE_PER_ROUND)
    reset_to_init(model_ticket, W0)
    train_one_run(model_ticket, epochs=TRAIN_EPOCHS, lr=1e-3)

# confusion matrix
all_preds, all_labels = [], []
model_ticket.eval()
with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        preds  = model_ticket(images).argmax(dim=1).cpu()
        all_preds.extend(preds.numpy())
        all_labels.extend(labels.numpy())

cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=CLASSES, yticklabels=CLASSES)
plt.title(f"Winning Ticket — Confusion Matrix  "
          f"(round {best_round['round']}, sparsity {best_round['sparsity']*100:.0f}%)")
plt.ylabel("True label")
plt.xlabel("Predicted label")
plt.tight_layout()
plt.show()

print("\nClassification report:")
print(classification_report(all_labels, all_preds, target_names=CLASSES))


# ── 10. PREDICTION VISUALIZATION ─────────────────────────────

test_images, test_labels_batch = next(iter(test_loader))
with torch.no_grad():
    outputs = model_ticket(test_images.to(device))
    _, preds = torch.max(outputs, 1)
    preds = preds.cpu()

fig, axes = plt.subplots(3, 6, figsize=(15, 8))
axes = axes.flatten()
for i in range(18):
    img  = test_images[i].squeeze()
    true = test_labels_batch[i].item()
    pred = preds[i].item()
    color = 'green' if true == pred else 'red'
    axes[i].imshow(img, cmap='gray')
    axes[i].set_title(f"T:{CLASSES[true][:3]}\nP:{CLASSES[pred][:3]}", color=color, fontsize=8)
    axes[i].axis('off')
plt.suptitle(f"Winning Ticket Predictions  (round {best_round['round']}, sparsity {best_round['sparsity']*100:.0f}%)")
plt.tight_layout()
plt.show()


# ── 11. SPARSITY PER LAYER ───────────────────────────────────

print("\nSparsity breakdown per layer (winning ticket):")
print(f"{'Layer':40s}  {'Total':>8}  {'Alive':>8}  {'Dead':>8}  {'Sparsity':>10}")
print("-" * 78)
for name, mask in model_ticket.masks.items():
    total  = mask.numel()
    alive  = (mask == 1).sum().item()
    dead   = total - alive
    sp     = dead / total * 100
    print(f"  {name:38s}  {total:>8,}  {alive:>8,}  {dead:>8,}  {sp:>9.1f}%")

total_all  = sum(m.numel()           for m in model_ticket.masks.values())
alive_all  = sum((m==1).sum().item() for m in model_ticket.masks.values())
print("-" * 78)
print(f"  {'TOTAL':38s}  {total_all:>8,}  {alive_all:>8,}  {total_all-alive_all:>8,}  {(total_all-alive_all)/total_all*100:>9.1f}%")
print(f"\nCompression: {total_all/alive_all:.1f}x fewer active parameters")
```
