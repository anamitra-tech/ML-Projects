```python
# =========================
# 1. IMPORTS
# =========================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import confusion_matrix, recall_score, f1_score


# =========================
# 2. LOAD DATA
# =========================
df = pd.read_csv('/content/customer_churn_buisness_dataset.csv')

# Standardize column names
df.columns = df.columns.str.lower().str.replace('_', ' ', regex=False)


# =========================
# 3. BASIC EDA
# =========================
print("Dataset Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTarget Distribution (Counts):")
print(df['churn'].value_counts())

print("\nTarget Distribution (%):")
print(df['churn'].value_counts(normalize=True) * 100)

plt.figure(figsize=(5,4))
sns.countplot(x='churn', data=df)
plt.title('Severely Imbalanced Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Customer Count')
plt.show()


# =========================
# 4. DATA CLEANING & FEATURE HANDLING
# =========================

# Binary categorical mapping
df['discount applied'] = df['discount applied'].map({'Yes': 1, 'No': 0})
df['price increase last 3m'] = df['price increase last 3m'].map({'Yes': 1, 'No': 0})

# Drop identifier (no predictive value)
df = df.drop(columns=['customer id'])

# Separate target
y = df['churn']
X = df.drop(columns=['churn'])

# Identify categorical columns
cat_cols = [
    'gender','country','city','customer segment','signup channel',
    'contract type','payment method','complaint type',
    'nps score','survey response'
]

# One-hot encode categoricals
X = pd.get_dummies(X, columns=cat_cols, drop_first=True)


# =========================
# 5. STRATIFIED K-FOLD SETUP
# =========================
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Recall-focused threshold (business-driven)
threshold = 0.40

# Class imbalance ratio for XGBoost
scale_pos = y.value_counts()[0] / y.value_counts()[1]


# =========================
# 6. MODEL DEFINITIONS
# =========================
models = {
    "Logistic Regression": Pipeline([
        ('scaler', StandardScaler()),
        ('lr', LogisticRegression(
            class_weight='balanced',
            max_iter=1000
        ))
    ]),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=6,
        class_weight='balanced',
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=8,
        class_weight='balanced',
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        n_estimators=350,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos,
        gamma=0.1,
        min_child_weight=1,
        reg_alpha=0.1,
        reg_lambda=0.1,
        objective='binary:logistic',
        eval_metric='logloss',
        random_state=42
    )
}


# =========================
# 7. STRATIFIED CV TRAINING + CONFUSION MATRICES
# =========================
results = []

for name, model in models.items():

    all_y_true = []
    all_y_pred = []

    for train_idx, test_idx in skf.split(X, y):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        model.fit(X_train, y_train)
        probs = model.predict_proba(X_test)[:, 1]
        preds = (probs >= threshold).astype(int)

        all_y_true.extend(y_test)
        all_y_pred.extend(preds)

    recall = recall_score(all_y_true, all_y_pred)
    f1 = f1_score(all_y_true, all_y_pred)

    results.append((name, recall, f1))

    cm = confusion_matrix(all_y_true, all_y_pred)

    plt.figure(figsize=(5,4))
    sns.heatmap(
        cm, annot=True, fmt='d', cmap='Blues',
        xticklabels=['Not Churn', 'Churn'],
        yticklabels=['Not Churn', 'Churn']
    )
    plt.title(f'Confusion Matrix (Stratified CV) – {name}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()


# =========================
# 8. FINAL COMPARISON
# =========================
results_df = pd.DataFrame(
    results, columns=['Model', 'Recall', 'F1 Score']
).sort_values(by='Recall', ascending=False)

print("\nModel Comparison (Recall & F1 using Stratified CV):")
print(results_df)
```
